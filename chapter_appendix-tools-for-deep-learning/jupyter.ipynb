```python
import numpy as np
from sklearn.linear_model import LinearRegression

#data set

X = np.array([
    [800,2],
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4]])
y = np.array([50, 60, 75, 90, 120])

# model
model = LinearRegression()
model.fit(X,y)


# prediction
prediction  = model.predict([[1600,3],[1700,3]])
print("predicted price is- ", prediction)
```
